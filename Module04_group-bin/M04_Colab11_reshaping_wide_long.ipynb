## Wide / summary Form Data
In real data, we might have a large number of measurements on each line. And the data are presented in wide form (aka summary form) like the one below - each row presents a summary of one year with 12 separate measurements.


````python
import pandas as pd
london_rain = pd.read_csv("https://raw.githubusercontent.com/csbfx/advpy122-data/master/london_rainfall.csv")
london_rain.head()
````

**Wide form** can be used to answer certain questions about the data. For example, the difference in rainfall between January and June and assign it to a new column for visualization (distribution plot).

````python
# london_rain["June"] - london_rain["January"]
````

We can store that as a new column in the dataframe and visualize it

````python
london_rain["January -> June difference"] = (
     london_rain["June"] - london_rain["January"]
)
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
sns.displot(london_rain["January -> June difference"], color="lightblue")
plt.title(
    """Distribution of difference in total rainfall
    between January and June in London""");
````

However, many other visualizations require the data to be in long-form or tidy form. With data in summary form, there is no way to use month as a category, so we can't draw charts that are colored by month, or draw a boxplot with month on the categorical axis.

### Question: How do we plot each months' rain value?

Melt is a function used to reshape a wide dataset into a long format. Wide data has multiple columns for different variables, while long data has just one column per variable. Melting is a way of "untangling" or collapsing multiple columns into key-value pairs, which makes the data long and typically easier to work with for statistical analysis.

## `melt()`
Reshape data from wide (summary) to long (tidy) form using `melt`. This function is useful to reshape a DataFrame into a format where a column(s) can function as the identifier variables using `id_vars=` and the remaining columns are the unpivot columns and specify via  `value_vars=`. If you do not specify `value_vars=`, then all columns not specified by id_vars are unpivot.

The Dataframe has two identifier columns that function as the variable and value. You can specify the name to use as the variable column using `var_name`. `value_name` is used to name the value column and cannot be an exisiting column label.

````python
tidy_rain = london_rain.melt(
    id_vars=
    var_name=
    value_name=
)
tidy_rain
````

Recall: a **tidy** dataset stores a unique observation per row. The dataframe "tidy_rain" above is a tidy dataset that has a single measurement per row.

Now, we can use the month as a category to create a categorical plot using seaborn.

````python
import numpy as np
np.sort(london_rain.Year.unique())
````

### Point plot
Here we use a `catplot()` to display point estimates (eg., mean) using dot positions using the parameter `kind="point"`. The points include error bars to show uncertainty with respect to the estimate.

We will also filter out the "January -> June difference" column

````python
### Plot point plot of monthly rainfall
g = sns.catplot(
    data=tidy_rain,
    x="Month",
    y="Rainfall (mm)",

)

g.fig.suptitle('''Mean total monthly rainfall in London
between 1960 and 2019''');
````

````python
### Another option to plot is to use kind="box"
g = sns.catplot(
    data = tidy_rain[tidy_rain["Month"] != "January -> June difference"],
    y="Month",
    x="Rainfall (mm)",

)
g.fig.suptitle('''Mean total monthly rainfall in London
between 1960 and 2019''', y=1.06);
````

## wide_to_long()
 In addition to `melt`, pandas also another function called `wide_to_long`. We can use pandas’ [`wide_to_long()`](https://pandas.pydata.org/docs/reference/api/pandas.wide_to_long.html) to reshape the wide dataframe into long/tall dataframe. Under the hood pandas `wide_to_long()` uses pandas’ `melt` function. However, Pandas’ wide_to_long() is easier to use. Another benefit of using pandas `wide_to_long()` is that we can easily take care of the *prefix* in the column names.

````python
### Import worldwide dataset of GDP
worldwide = pd.read_csv("https://raw.githubusercontent.com/csbfx/advpy122-data/master/worldwide_data.csv")
worldwide.head()
````

````python
### Examine the columns to observe if there is a common prefix
worldwide.columns
````

### Working with prefix in wide_to_long()
We need to specify "stubnames" to extract the prefix from column variable names. In our example, `stubnames=['lifeExp','gdpPercap','pop']`. We also need to specify how the column names are separated `sep='_'`. We also need to specify a column that is unique to each row. In our example, `i='country'` is unique to each row. `i=` is used to identify the column(s) to use as id variables.  And finally, we need to specify the new variable name that we wish to use in the suffix in the long format by using the parameter `j=`; In our example `j='year'`.

````python
world_tidy = pd.wide_to_long(worldwide,
                             stubnames= ,
                             i= ,
                              j= ,
                             sep= )
world_tidy
````

## CategoricalDtype

````python
weather = pd.read_csv("https://raw.githubusercontent.com/csbfx/advpy122-data/master/weather.csv")
weather
````

````python
weather.info()
````

````python
mean_temp = weather.groupby(["City","Month"])["Mean temperature"].mean().reset_index()
mean_temp
````

````python
mean_temp.info()
````

Let's plot the mean temerature by month for all three cities.

````python
sns.catplot(
    data=mean_temp,
    y="Month",
    x="Mean temperature",
    hue="City",
    kind="bar"
)
plt.title("Mean temperature for each city by month");
````

Let's order the months in chronological order.

````python
months=["January", "February", "March", "April", "May",\
        "June", "July", "August", "September", "October",\
        "November", "December"]

sns.catplot(
    data=mean_temp,
    y="Month",
    x="Mean temperature",
    hue="City",
    kind="bar",
    order=months
)
plt.title("Mean temperature for each city by month");
````

Sometimes we want the a column to use a datatype to follow a specific sort order. For instance, from January to Decemember for `Month`. We can use the module `CategoricalDtype` from `pandas.api.types`

````python
from pandas.api.types import CategoricalDtype

months=["January", "February", "March", "April", "May",\
        "June", "July", "August", "September", "October",\
        "November", "December"]

mean_temp["Month"] = mean_temp["Month"].astype(
    CategoricalDtype(categories=months, ordered=True)
)
````

````python
mean_temp.dtypes
````

````python
sns.catplot(
    data=mean_temp,
    y="Month",
    x="Mean temperature",
    hue="City",
    kind="bar"
)
plt.title("Mean temperature for each city by month");
````

## Reshape dataframe from long (tidy) to wide (summary) using `pivot()`
Sometime we may want to take data frame with fewer columns, say in long format, summarize and convert into a data frame with multiple columns, i.e. a wide data frame. We can use pandas function `pivot_table` to help us with the summarization and conversion of dataframe in long form to dataframe in wide form.

````python
world = pd.read_csv("https://raw.githubusercontent.com/csbfx/advpy122-data/master/world_data.csv")
world
````

Let us see a simple example of python `pivot_table` using a dataframe with just two columns. Let us subset our dataframe to contain just two columns: `continent` and `lifeExp`.

````python
# simple example with pivot_table
pd.pivot_table(world, values='lifeExp', columns='continent')
````

````python
# simple example with pivot_table
pd.pivot_table(world, values=['lifeExp','gdpPercap'], columns='continent')
````

Here we use `pivot_table` to convert the long dataframe to a wide dataframe, and computes the mean lifeExp across continents. `pd.pivot_table` with the dataframe `world` as one of the arguments and specify which variable we would like use for columns (`continent`) and which variable we would like to summarize (`lifeExp`). One of the arguments of `pivot_table`, `aggfunc` has mean as default.

In the above example we used `pivot_table` to compute mean `lifeExp` for each continent. We can compute mean lifeExp for each country, by simply specifying "country" for columns argument.

````python
country = world[['country','lifeExp']]
country
````

## Pandas `pivot_table` on dataframes with three columns
Pandas `pivot_table` becomes more powerful when we try to summarize and convert a tidy dataframe with more than two variables into a wide dataframe.

````python
world3 = world[['continent', 'year','lifeExp']]
world3
````

````python
# pivot table example with three columns
pd.pivot_table(world3, values='lifeExp',
                    index=['year'],
                    columns='continent',
                    aggfunc='min'
               )
````

## `pivot_table` with Different Aggregating Function
`pivot_table` uses mean function for aggregating or summarizing data by default. We can change the aggregating function, if needed. For example, we can use `aggfunc='min'` to compute "minimum" lifeExp instead of "mean" lifeExp for each year and continent values.

````python
pd.pivot_table(world3, values='lifeExp',
                    index=['year'],
                    columns='continent',
                    aggfunc='min')
````

We can also specify more than one aggregating functions. For example, if we are interested in bot minimum and maximum values of lifeExp for each year and continent, we can specify the functions as a list to the argument "aggfunc".

````python
pd.pivot_table(world3, values='lifeExp',
                     index=['year'],
                     columns='continent',
                     aggfunc=["min","max","mean"])
````

When we give multiple aggregating functions, we would get a multi-indexed data frame as output.
