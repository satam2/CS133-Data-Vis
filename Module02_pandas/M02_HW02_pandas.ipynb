#### Question 6: ####
Create a variable `adjusted_price` containing the adjusted price which is the price subtracted by the average price.

*This is called **"centering" transformation** - a method commonly used in the preprocessing step before applying various machine learning algorithms.*

```python
mean_price = wine["price"].mean()
adjusted_price = wine["price"].map(lambda price: price - mean_price)
adjusted_price
```

#### Question 9: ####
Use `apply()` with `axis=1` to create a function called `is_bargain_func` to run a boolean indexing that is `True` when a wine's `price` is less than 15 and its `points` is greater than 85, and `False` otherwise. Use apply() to pass your function and return output to new column `is_bargain`.

```python
def is_bargain_func(row):
    return (row["price"] < 15) and (row["points"] > 85)


wine["is_bargain"] = wine.apply(is_bargain_func, axis=1)
wine[["price", "points", "is_bargain"]].head()
```

Alternative, you can use a **lamdba** function because we need to consider more than one column of the row at once. This requirement is why `.map()` alone is not sufficient.

```python
wine["is_bargain"] = wine.apply(
    lambda row: (row["price"] < 15) and (row["points"] > 85),
    axis=1
)
wine[["price", "points", "is_bargain"]].head()
```

#### Question 10: ####
Create a Series `flavor_counts` that contains two values: the number of wines that has the word "tart" in the `description` column and the number of wines that has the word "berries" in the `description` column. The index of the Series should be "Tart" and "Berries" for the corresponding values.

```python
# Count each description once per lowercase search term using map().
tart_count = wine["description"].map(lambda description: "tart" in description).sum()
berries_count = wine["description"].map(lambda description: "berries" in description).sum()

flavor_counts = pd.Series([tart_count, berries_count], index=["Tart", "Berries"])
flavor_counts
```

#### Question 11: ####
Let's convert the points into simple star ratings.
*   A score of 90 or higher counts as 3 stars, a score of at least 80 but less than 90 is 2 stars
*   Any other score is 1 star
*   Any wines from France should automatically get 3 stars, regardless of points.

Add this new column `star_ratings` to the dataframe with the number of stars for each wine in the dataset.

```python
def star_rating_func(row):
    if row["country"] == "France" or row["points"] >= 90:
        return 3
    elif row["points"] >= 80:
        return 2
    else:
        return 1


wine["star_ratings"] = wine.apply(star_rating_func, axis=1)
wine[["country", "points", "star_ratings"]].head()
```

#### Question 12: ####
What combination of countries and varieties are most common? Create a Series whose index is a MultiIndexof {country, variety} pairs. For example, a pinot noir produced in the US should map to {"US", "Pinot Noir"}. Sort the values in the Series in descending order based on wine count.
